```python
datasets,info=tfds.load("imdb_reviews",as_supervised=True,with_info=True)
train_size=info.splits["train"].num_examples
```

```python
def preprocess(x_batch, y_batch):
    x_batch = tf.strings.substr(x_batch, 0, 300)
    x_batch = tf.strings.regex_replace(x_batch, r"<br\s*/?>", " ")
    x_batch = tf.strings.regex_replace(x_batch, r"[^a-zA-Z']", " ")
    x_batch = tf.strings.split(x_batch)

    return x_batch.to_tensor(default_value=b"<pad>"), y_batch
```

```python
vocab=Counter()
for x_batch,y_batch in datasets["train"].batch(32).map(preprocess):
    for review in x_batch:
        vocab.update(list(review.numpy()))
vocab.most_common()[:3]
```

```python
vocab_size=1000
trunc_vocab=[word for word,count in vocab.most_common()[:vocab_size]]
trunc_vocab
```

```python
words= tf.constant(trunc_vocab)
word_ids=tf.range(len(trunc_vocab),dtype=tf.int64)
vocab_init=tf.lookup.KeyValueTensorInitializer(words,word_ids)
num_oov_buckets=1000
table=tf.lookup.StaticVocabularyTable(vocab_init,num_oov_buckets)
```

```python
table.lookup(tf.constant([b"This movie was faaaaantastic".split()]))
```

```python
def encoder_words(x_batch, y_batch):
    return table.lookup(x_batch), y_batch
```

```python
train_set=datasets["train"].batch(32).map(preprocess)
train_set=train_set.map(encoder_words).prefetch(1)
```

```python
embed_size=128
model=tf.keras.models.Sequential([
    tf.keras.layers.Embedding(vocab_size+num_oov_buckets,embed_size,input_shape=[None]),
    tf.keras.layers.GRU(128,return_sequences=True),
    tf.keras.layers.GRU(128),
    tf.keras.layers.Dense(1,activation="sigmoid")
])
```

```python
model.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)
```

```python
model.fit(train_set,epochs=5)
```
